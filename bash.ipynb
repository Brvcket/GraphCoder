## Precomputing embeddings for text based search

```python
def precompute_repo_embeddings(repo_name: str, embedder, embedder_name, force: bool = False):
    """
    Reads <graph_database>/<repo_name>.jsonl,
    for each case uses key_forward_encoding (if non-empty) else key_forward_context
    to compute case["embedding"], then writes <repo_name>.with_emb.jsonl.
    """
    in_path  = os.path.join(CONSTANTS.graph_database_save_dir, f"{repo_name}.jsonl")
    out_path = os.path.join(CONSTANTS.graph_database_save_dir, f"{repo_name}.{embedder_name}.with_emb.jsonl")

    if os.path.exists(out_path) and not force:
        print(f"[precompute] {out_path} exists; skipping (use --force).")
        return

    cases = load_jsonl(in_path)
    for case in cases:
        data   = case.get("key_forward_context", "")
        case["embedding"] = embedder.get_embedding(data).tolist()

    dump_jsonl(cases, out_path)
    print(f"[precompute] saved embeddings to {out_path}")
```

## Grid search

```python
import os
import itertools
import json
import torch
from tqdm import tqdm
import pandas as pd

from utils.utils import (
    load_jsonl, dump_jsonl, make_needed_dir,
    CodexTokenizer, CodeGenTokenizer, CONSTANTS
)
from utils.metrics import hit, compute_identifier_match, compute_EM, compute_ES
from transformers import AutoTokenizer, AutoModelForCausalLM, GenerationConfig
from search_code_2 import CodeSearchWorker
from utils.build_prompt import build_prompt

# -------------------------
# Configuration Grid
# -------------------------
EMBEDDERS = ['codet5', 'codebert', 'graphcodebert']
METRICS = ['cosine', 'dot', 'l2', 'l1']
MODES = ['embed_token', 'embed_embed'] #'token_token', 'token_embed', 
GAMMAS = [0.1]  # add more gamma values as desired
QUERY_CASES = 'api_level'
QFILE = f"{QUERY_CASES}.python.test.jsonl"

# limits
CASE_LIMIT = 100
GEN_LIMIT  = 100
```

```python
import torch
torch.cuda.empty_cache()
```

```python
# generation model setup
# model = AutoModelForCausalLM.from_pretrained("Qwen/Qwen2.5-Coder-3B-Instruct", trust_remote_code=True).to('cuda')
# tokenizer_raw = AutoTokenizer.from_pretrained("Qwen/Qwen2.5-Coder-3B-Instruct", trust_remote_code=True)

DEVICE = 'cuda' if torch.cuda.is_available() else 'cpu'
GEN_MODEL_NAME = 'Qwen/Qwen2.5-Coder-3B-Instruct'
MODEL = AutoModelForCausalLM.from_pretrained(GEN_MODEL_NAME, trust_remote_code=True, torch_dtype=torch.float16).to(DEVICE)
TOKENIZER_RAW = AutoTokenizer.from_pretrained(GEN_MODEL_NAME, trust_remote_code=True)
```

```python
# generation config
import time

gen_cfg = GenerationConfig(
    max_new_tokens=100,
    do_sample=False,
    eos_token_id=TOKENIZER_RAW.eos_token_id,
    pad_token_id=TOKENIZER_RAW.pad_token_id,
    temperature=0.0
)

# load targets
TARGET_PATH = os.path.join('RepoEval-Updated', f"{QUERY_CASES}.python.test.jsonl")
TARGET = load_jsonl(TARGET_PATH)

# prepare output paths
out_dir = 'grid_results'
os.makedirs(out_dir, exist_ok=True)
jsonl_path = os.path.join(out_dir, 'full_grid.jsonl')
csv_path = os.path.join(out_dir, 'full_grid.csv')

if not os.path.exists(csv_path):
    pd.DataFrame(columns=[
        'embedder','metric','mode','gamma',
        'hit1','hit5','hit10',
        'retrieval_time','generation_time',
        'id_em','id_f1','code_em','code_es',
        'search_out','gen_out'
    ]).to_csv(csv_path, index=False)
open(jsonl_path, 'a').close()

# -------------------------
# Run Full Grid Search
# -------------------------

def run_grid():
    for embedder, metric, mode, gamma in itertools.product(EMBEDDERS, METRICS, MODES, GAMMAS):
        tag = f"{embedder}.{metric}.{mode}.g{int(gamma*100)}"
        print(f"Running config: {tag}")

        qpath = os.path.join(CONSTANTS.query_graph_save_dir, embedder, QFILE)
        if not os.path.exists(qpath):
            raise FileNotFoundError(f"Queries not found at {qpath}")
        cases = load_jsonl(qpath)[:CASE_LIMIT]

        search_out = os.path.join('search_results', f"full.test.{tag}.jsonl")
        if os.path.exists(search_out):
            print(f"Retrieval output exists, skipping: {search_out}")
            retrieval_time = None
        else:
            print("Running retrieval...")
            make_needed_dir(search_out)
            t0 = time.time()
            CodeSearchWorker(
                query_cases=cases,
                output_path=search_out,
                mode=mode,
                gamma=gamma,
                embedder_choice=embedder,
                similarity_metric=metric
            ).run()
            retrieval_time = time.time() - t0

        search_res = load_jsonl(search_out)
        h1, h5, h10 = hit(search_res, hits=[1,5,10])

        gen_out = os.path.join('generation_results', f"gen.test.{tag}.jsonl")
        if os.path.exists(gen_out):
            print(f"Generation output exists, skipping: {gen_out}")
            responses = load_jsonl(gen_out)
            generation_time = None
        else:
            print("Running generation...")
            gen_input = load_jsonl(search_out)[:GEN_LIMIT]
            make_needed_dir(gen_out)
            responses = []
            codegen_tok = CodeGenTokenizer(TOKENIZER_RAW)

            t0g = time.time()
            for case in tqdm(gen_input, desc=f"Gen {tag}"):
                prompt = build_prompt(
                    case, codegen_tok,
                    max_num_tokens=2048-100,
                    max_top_k=len(case['top_k_context']),
                    mode='retrieval'
                )
                inputs = TOKENIZER_RAW(prompt, return_tensors='pt').to(DEVICE)
                out_ids = MODEL.generate(
                    inputs['input_ids'],
                    attention_mask=inputs['attention_mask'],
                    generation_config=gen_cfg
                )
                full = codegen_tok.decode(out_ids[0])
                resp = full[len(prompt):]
                record = dict(case)
                record['generate_response'] = resp
                responses.append(record)
            dump_jsonl(responses, gen_out)
            generation_time = time.time() - t0g

        subset = TARGET[:len(responses)]
        id_em, id_f1, code_em, code_es = 0, 0, 0, 0
        for pred, truth in zip(responses, subset):
            gt, hyp = truth['metadata']['ground_truth'], pred['generate_response']
            em_i, f1_i = compute_identifier_match(hyp, gt, language='python')
            code_em += compute_EM(gt, hyp, language='python')
            code_es += compute_ES(gt, hyp, language='python')
            id_em += em_i
            id_f1 += f1_i

        n = len(responses)
        result = {
            'embedder': embedder, 'metric': metric, 'mode': mode, 'gamma': gamma,
            'hit1': h1, 'hit5': h5, 'hit10': h10,
            'retrieval_time': retrieval_time,
            'generation_time': generation_time,
            'id_em': id_em/n, 'id_f1': id_f1/n,
            'code_em': code_em/n, 'code_es': code_es/n,
            'search_out': search_out,
            'gen_out': gen_out
        }

        with open(jsonl_path, 'a') as jf:
            jf.write(json.dumps(result) + '\n')
        pd.DataFrame([result]).to_csv(csv_path, mode='a', header=False, index=False)

    print("Grid search complete!")
    print(f"Results saved to:\n  {jsonl_path}\n  {csv_path}")

if __name__ == '__main__':
    run_grid()
```

```python
model = AutoModelForCausalLM.from_pretrained("Qwen/Qwen2.5-Coder-3B-Instruct", trust_remote_code=True).to('cuda')
tokenizer_raw = AutoTokenizer.from_pretrained("Qwen/Qwen2.5-Coder-3B-Instruct", trust_remote_code=True)
```

```python
def generate(model, tokenizer_raw, mode):
    tokenizer = CodeGenTokenizer(tokenizer_raw)
    
    max_num_tokens = 2048
    max_new_tokens = 100
    max_top_k = 10
    
    generation_config = GenerationConfig(
        max_new_tokens=max_new_tokens,
        do_sample=False,
        eos_token_id=tokenizer_raw.eos_token_id,
        temperature=0,
        pad_token_id=tokenizer_raw.pad_token_id,
    )
    responses = []
    max_prompt_tokens = max_num_tokens - max_new_tokens
    with tqdm(total=len(input_cases)) as pbar:
        for case in input_cases:
            pbar.set_description(f'Processing...')
     
            prompt = build_prompt(case, tokenizer, max_prompt_tokens, max_top_k=max_top_k, mode=mode)
            prompt_ids = tokenizer_raw(prompt, return_tensors="pt").to('cuda')
            response_ids = model.generate(prompt_ids['input_ids'],
                                          generation_config=generation_config,
                                          attention_mask=prompt_ids['attention_mask'])
            response = tokenizer.decode(response_ids[0])
            prompt_lines = prompt.splitlines(keepends=True)
            n_prompt_lines = len(prompt_lines)
            response_lines = response.splitlines(keepends=True)
            response = "".join(response_lines[n_prompt_lines:])
            case_res = copy.deepcopy(case)
            case_res['generate_response'] = response
            responses.append(case_res)
            pbar.update(1)
    dump_jsonl(responses, responses_save_name)
```

```python
generate(model, tokenizer_raw, "retrieval")
```

```python
from utils.metrics import compute_identifier_match, compute_EM, compute_ES
from utils.utils import load_jsonl

def evaluate(target, prediction):
    em_val = 0
    f1_val = 0
    code_em_val = 0
    code_es_val = 0
    for i in range(0, len(prediction)):
        pred_case = prediction[i]
        pred_str = pred_case['generate_response']
        gt_str = target[i]['metadata']['ground_truth']
        
        em, f1 = compute_identifier_match(pred_str, gt_str, language="python")
        em_val += em
        f1_val += f1
    
        code_em = compute_EM(gt_str, pred_str, language="python")
        code_es = compute_ES(gt_str, pred_str, language="python")
        code_em_val += code_em
        code_es_val += code_es
        
    return code_em_val/len(prediction), code_es_val/len(prediction), em_val / len(prediction), f1_val/len(prediction)
```
